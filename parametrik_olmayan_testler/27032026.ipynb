## İşaret Testleri

```r
library(BSDA)
veri <- data.frame(
    once=c(58, 62, 70, 66, 71, 69, 64, 73, 67, 61, 68, 72),
    sonra=c(60, 65, 72, 68, 70, 71, 66, 74, 69, 63, 69, 75)
)
print(veri)
cat("\n")
```

```r
# Yardımcı fonksiyon: P-değerine göre hipotez test sonucunu yorumlar.
hypothesis_test <- function(pvalue, alpha=0.05) {
    if (pvalue < alpha) {
        return("H0 reddedilir, H1 kabul edilir.")
    } else {
        return("H0 kabul edilir, H1 reddedilir.")
    }
}
```

```r
# 
# Önce değişkeninin medyan değerinin 70'den farklı olup olmadığını test eder.
# 
sign_result_once = SIGN.test(
    veri$once,
    md=70
)

cat("Hipotezler | Önce Değişkeni İşaret Testi\n")
cat("H0: Önce değişkeninin medyan değerinin 70'den farkı yoktur.\n")
cat("H1: Önce değişkeninin medyan değeri 70'den farklıdır.\n")
cat("P-değeri:", sign_result_once$p.value, "\n")
cat("Hipotez Sonucu:", hypothesis_test(sign_result_once$p.value), "\n")
cat("\n\n")
```

```r
# 
# Önce ve Sonra değişkenlerinin medyan değerleri arasında fark olup olmadığını test eder.
# 
sign_result_sonra = SIGN.test(
    veri$once,
    veri$sonra,
)

cat("Hipotezler | Önce ve Sonra Değişkenleri İşaret Testi\n")
cat("H0: Önce ve sonra değişkenlerinin medyan değerleri arasında fark yoktur.\n")
cat("H1: Önce ve sonra değişkenlerinin medyan değerleri arasında fark vardır.\n")
cat("P-değeri:", sign_result_sonra$p.value, "\n")
cat("Hipotez Sonucu:", hypothesis_test(sign_result_sonra$p.value), "\n")
cat("\n\n")
```

```r
# 
# Önce değişkeninin medyan değerinin 70'den farklı olup olmadığını Wilcoxon testi ile test eder.
#
sign_result_wilcox = wilcox.test(
    veri$once,
    mu=70
)
cat("Hipotezler | Önce Değişkeni Wilcoxon Testi\n")
cat("H0: Önce değişkeninin medyan değerinin 70'den farkı yoktur.\n")
cat("H1: Önce değişkeninin medyan değeri 70'den farklıdır.\n")
cat("P-değeri:", sign_result_wilcox$p.value, "\n")
cat("Hipotez Sonucu:", hypothesis_test(sign_result_wilcox$p.value), "\n")
```

## Bağımlı Kategorik Değişkenler için McNemar Testi

```r
library(BSDA)
veri <- data.frame(
    once=c("Evet", "Hayır", "Hayır", "Evet", "Evet", "Evet"),
    sonra=c("Evet", "Evet", "Hayır", "Hayır", "Hayır", "Evet") 
)
```

```r
veri$once <- factor(veri$once, levels=c("Evet", "Hayır"))
veri$sonra <- factor(veri$sonra, levels=c("Evet", "Hayır"))
```

```r
cabtab <- table(veri$once, veri$sonra)
cabtab
```

```r
mcnemar_result <- mcnemar.test(veri$once, veri$sonra, correct=TRUE)
mcnemar_result
```

```r
cat("Hipotezler | McNemar Testi\n")
cat("H0: Önce ve sonra değişkenleri arasında fark yoktur.\n")
cat("H1: Önce ve sonra değişkenleri arasında fark vardır.\n")
cat("P-değeri:", mcnemar_result$p.value, "\n")
cat("Hipotez Sonucu:", hypothesis_test(mcnemar_result$p.value), "\n")
```
