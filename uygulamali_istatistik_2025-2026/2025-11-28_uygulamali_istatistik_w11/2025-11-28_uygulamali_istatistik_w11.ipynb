### Tek Örneklem için Wilcoxon İşaret Testi

```r
# install.packages("signmedian.test")
# library(signmedian.test)
# signmedian.test(zaman, mu = 20, alternative="two.sided", exact=FALSE)
zaman <- c(14, 26, 25, 10, 29, 26, 26, 16, 9, 9, 27, 5, 11, 28, 22, 22, 10, 26, 23, 22)
wilcox.test(zaman, mu = 20, alternative="greater")
```

### Bağımsız Örneklemler için Mann-Whitney U Testi

Veri


| Sıra Numarası | Erkek | Kadın |
|---------------|-------|-------|
| 1             | 49    | 45    |
| 2             | 41    | 39    |
| 3             | 35    | 50    |
| 4             | 46    | 48    |
| 5             | 58    | 34    |
| 6             | 48    | 29    |
| 7             | 35    | 49    |
| 8             | 45    | 44    |
| 9             | 57    | 42    |
| 10            | 37    | 42    |

```r
erkek <- c(49, 41, 35, 46, 58, 48, 35, 45, 57, 37)
kadin <- c(45, 39, 50, 48, 34, 29, 49, 44, 42, 42)
veri <- c(erkek, kadin)
cinsiyet <- rep(c("erkek", "kadin"), each=10)
wilcox.test(veri~cinsiyet)
```

### Bağımlı(Eşleştirilmiş) Örneklemler için Wilcoxon İşaretli Sıralar Testi

Veri


| Satış Rakamları | Nisan | Mayıs |
|-----------------|-------|-------|
| 1               | 15    | 20    |
| 2               | 23    | 22    |
| 3               | 17    | 18    |
| 4               | 18    | 19    |
| 5               | 20    | 17    |
| 6               | 21    | 19    |
| 7               | 19    | 20    |

```r
nisan <- c(15, 23, 17, 18, 20, 21, 19)
mayis <- c(20, 22, 18, 19, 17, 19, 20)
wilcox.test(nisan, mayis)
```

### Varyans Analizi için Kruskal-Wallis H Testi

```r
kruskal.test(chickwts$weight ~ chickwts$feed)
kruskal.test(weight ~ feed, data = chickwts)
```
