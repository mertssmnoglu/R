Korelasyon katsayısı **r** ile gösterilir ve -1 ile +1 arasında bir değer alır.
Testin ilişkisini, gücünü ve yönünü belirler.


| Korelasyon Katsayısı (r) | Yorum               |
| ------------------------ | ------------------- |
| 0 - 0.05                 | İlişki yok          |
| 0.06 - 0.19              | Çok zayıf ilişki    |
| 0.20 - 0.39              | Zayıf ilişki        |
| 0.40 - 0.59              | Orta düzeyde ilişki |
| 0.60 - 0.79              | Güçlü ilişki        |
| 0.80 - 1.0               | Çok güçlü ilişki    |

Eğer r pozitif ise, değişkenler arasındaki ilişki pozitif yönlüdür (biri artarken diğeri de artar).\
Eğer r negatif ise, ilişki negatif yönlüdür (biri artarken diğeri azalır).

## Korelasyon Türleri

### Pearson Korelasyonu

Pearson korelasyonu, iki sürekli değişken arasındaki doğrusal ilişkiyi ölçer.

### Spearman Korelasyonu

Spearman korelasyonu, iki sıralı değişken arasındaki ilişkiyi ölçer ve verilerin normal dağılım göstermediği durumlarda kullanılır.

### Kendall Tau Korelasyonu

Kendall Tau korelasyonu, sıralı veriler arasındaki ilişkiyi ölçer ve özellikle küçük örneklemler için uygundur.

## Uygulama

````r
# Virgülle ayrılmış dosya(CSV) ile veri yükleme
veri <- read.csv("veri.csv")

# Tab ile ayrılmış dosya(TSV) ile veri yükleme, (.txt uzantılı olabilir)
# veri <- read.table("veri.txt", header = TRUE, sep = "\t")

plot(veri)
with(veri, abline(lm(Y~X), col="dark blue"))
````

````r
ks.test(veri$X, veri$Y)
cat("Korelasyon Katsayısı (Pearson): ", cor(veri$X, veri$Y, method = "pearson"), "\n")
cat("---")
cor.test(veri$X, veri$Y, method = "pearson")
cat("---")
cor.test(veri$X, veri$Y, method = "spearman")
cat("---")
cor.test(veri$X, veri$Y, method = "kendall")
````

## Hipotezler

H0: Değişkenler arasında ilişki yoktur. (r = 0)\
H1: Değişkenler arasında ilişki vardır. (r ≠ 0)
