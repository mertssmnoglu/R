```r
cars
speed<-cars$speed
dist<-cars$dist
plot(speed,dist,xlim=c(0,150),ylim=c(0,150),xlab="SPEED",ylab="DIST",main="REGRESYON",pch=19)
regresyon<- lm(dist~speed)
abline(regresyon,col="red")
summary(regresyon)
ozet<- summary(regresyon)
```

```r
# �nemli verilerin elde edilmesi
ozet$r.squared
ozet$adj.r.squared
sqrt(ozet$r.squared)
confint(regresyon)
coef(regresyon)

fitted(regresyon)
fit<-fitted(regresyon)
predict(regresyon,newdata=data.frame(speed=c(30,35,40)))
resid(regresyon)
artik<-resid(regresyon)
summary(artik)
plot(scale(artik))  
abline(h=2, col="red")
abline(h=-2, col="red")

anova(regresyon)
```

```r
# install.packages("lmtest")
library(lmtest)

# Ayk�r� de�er bulummas�
plot(cooks.distance(regresyon))

# S�f�r Ortalama Varsay�m�
mean(residuals(regresyon))

# Sabit varyans varsay�m�
# bptest(regresyon)
library(car)
ncvTest(regresyon)

plot(scale(artik))  
abline(h=2, col="red")
abline(h=-2, col="red")

# Normallik varsay�m�
shapiro.test(artik)
hist(artik)
qqplot(speed,artik)

# Otokorelasyon
dwtest(regresyon, alternative = "two.sided")
```

```r
# install.packages("jtools")
library(jtools)
trees

Girth<-trees$Girth
Height<-trees$Height
Volume<-trees$Volume
cor(trees)
plot(trees,pch=19,col="RED",main="VOLUME-GRITH-HEIGHT (Scatterp-lot)")

lm(formula = trees$Volume ~ trees$Girth + trees$Height)
regresyon<- lm(formula = trees$Volume ~ trees$Girth + trees$Height)
summary(regresyon)
ozet<- summary(regresyon)
fitted(regresyon)
newdata=data.frame(Girth=c(10,15,20), Height=c(68,73,78))
predict(regresyon, newdata)
resid(regresyon)
artik<-resid(regresyon)
summary(artik)
plot(scale(artik))  
abline(h=2, col="red")
abline(h=-2, col="red")

anova(regresyon)

# if (require(HH)) install.packages("HH")
# library(HH)
# ci.plot(regresyon)

# Ayk�r� de�er bulummas�
plot(cooks.distance(regresyon))

# S�f�r Ortalama Varsay�m�
mean(residuals(regresyon))

# Sabit varyans varsay�m�
# https://cran.r-project.org/web/packages/AFR/vignettes/Diagnostic-tests.html
# bptest(regresyon)

plot(scale(artik))  
abline(h=2, col="red")
abline(h=-2, col="red")

# Normallik varsay�m�
shapiro.test(artik)
hist(artik)
qqplot(Volume,artik)

# Otokorelasyon
#https://www.geeksforgeeks.org/r-language/understanding-durbin-watson-test-in-r/
# https://search.r-project.org/CRAN/refmans/lmtest/html/dwtest.html
# https://rpubs.com/DragonflyStats/Durbin-Watson-Test
#install.packages("lmtest")
library(lmtest)
dwtest(regresyon, alternative = "two.sided")

library(car)
durbinWatsonTest(regresyon)

# �oklu ba�lant�
cor(trees)
library(car)
vif(regresyon) 
sqrt(vif(regresyon))
sqrt(vif(regresyon))>2

#https://statsandr.com/blog/multiple-linear-regression-made-simple/
# jtools paketi ile
summ(regresyon)
```
