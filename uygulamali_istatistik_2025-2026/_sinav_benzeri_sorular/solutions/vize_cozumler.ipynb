# Sınav Benzeri Vize Çözümleri: Uygulamalı İstatistik 2025-2026


a) `boy` vektörünü oluşturun ve görüntüleyin.


```r
boy <- c(1, 3, 2, 5, 6, 7, 12, 34, 56, 78, 90, 123)
boy
```

b) `boy` için ortalama, medyan, standart sapma ve değişim aralığını hesaplayın.


```r
ortalama_boy <- mean(boy)
medyan_boy <- median(boy)
ssapma_boy <- sd(boy)
aralik_boy <- max(boy) - min(boy)
ortalama_boy
medyan_boy
ssapma_boy
aralik_boy
```

c) `boy` değişkenine ait özet değerleri raporlayın.


```r
summary(boy)
```

## Soru 2: Kategorik veriyi görselleştirme


a) `mtcars` veri çerçevesini çağırın ve ilk gözlemleri inceleyin.


```r
data(mtcars)
head(mtcars)
```

b) `am` değişkenini anlamlı etiketler taşıyan bir faktöre dönüştürün.


```r
am_faktor <- factor(mtcars$am, labels = c("Otomatik", "Manuel"))
am_faktor
```

c) `cyl` ve dönüştürülmüş `am` bilgilerini kullanarak yan yana gösterilen bir çubuk grafiği çizin.


```r
frekans_tablo <- table(mtcars$cyl, am_faktor)
frekans_tablo
barplot(
  frekans_tablo,
  beside = TRUE,
  col = c("royalblue", "darkorange", "seagreen"),
  legend.text = rownames(frekans_tablo),
  args.legend = list(title = "Silindir", x = "topright"),
  main = "Silindir ve Vites Türü Dağılımı",
  xlab = "Vites türü",
  ylab = "Frekans"
)
```

## Soru 3: Normal dağılım olasılıkları


a) P(X < 16) değerini hesaplayın.


```r
ortalama <- 15
sapma <- 3
p_kucuk_16 <- pnorm(16, mean = ortalama, sd = sapma)
p_kucuk_16
```

b) P(X > 18) değerini hesaplayın.


```r
p_buyuk_18 <- pnorm(18, mean = ortalama, sd = sapma, lower.tail = FALSE)
p_buyuk_18
```

c) P(6 < X < 19) değerini hesaplayın ve standart normal yoğunluk eğrisi üzerinde -1 ile 0.5 arasındaki alanı renklendiren bir grafik üretin.


```r
p_aralik <- pnorm(19, mean = ortalama, sd = sapma) - pnorm(6, mean = ortalama, sd = sapma)
p_aralik
z_degerleri <- seq(-4, 4, length.out = 200)
yogunluk <- dnorm(z_degerleri)
plot(
  z_degerleri,
  yogunluk,
  type = "l",
  xlab = "Z değeri",
  ylab = "Yoğunluk",
  main = "Standart Normal Yoğunluk Eğrisi"
)
golge_x <- seq(-1, 0.5, length.out = 200)
polygon(
  c(-1, golge_x, 0.5),
  c(0, dnorm(golge_x), 0),
  col = "skyblue",
  border = NA
)
```
