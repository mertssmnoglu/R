10. Uniform fonksiyonu ile 1 ile 100 arasında 1000 sayı üreten ve bu sayıları 5 gruba bölerek farklı renklerde ve frekansları sütun üstünde gösterecek sütun grafiğini çizen R kodunu yazınız.

```r
# ❌ Bu konu derste işlenmedi; referans amaçlı örnek kod.
set.seed(123)
uniform_values <- runif(1000, min = 1, max = 100)
grup_araligi <- seq(0, 100, length.out = 6)
uniform_gruplari <- cut(uniform_values, breaks = grup_araligi, include.lowest = TRUE)
frekanslar <- table(uniform_gruplari)
bar_renkleri <- c("navy", "steelblue", "lightblue", "gold", "tomato")
bar_orta_noktalari <- barplot(
  frekanslar,
  col = bar_renkleri,
  ylim = c(0, max(frekanslar) * 1.15),
  main = "Uniform(1,100) Değerlerinin 5 Gruplu Dağılımı",
  xlab = "Değer Aralığı",
  ylab = "Frekans"
)
text(bar_orta_noktalari, frekanslar, labels = frekanslar, pos = 3, cex = 0.8)
```

```r
set.seed(456)
norm_values <- rnorm(100, mean = 12, sd = 3)
x <- seq(min(norm_values), max(norm_values), length.out = 200)
y <- dnorm(x, mean = mean(norm_values), sd = sd(norm_values))
hist(
  norm_values,
  probability = TRUE,
  col = "mistyrose",
  border = "white",
  main = "Normal(12,3) Örneklem Yoğunluğu",
  xlab = "x"
)
lines(x, y, col = "darkorange", lwd = 2)
```

22. Hipotez testleri hakkında bilgi veriniz.


- Hipotez testleri, parametrelerle ilgili iddiaları veri desteğiyle değerlendirmek için kullanılır.
- İlk adımda H0 (sıfır) ve H1 (alternatif) hipotezleri kurulur ve anlamlılık düzeyi seçilir.
- Uygun test istatistiği hesaplanır, ardından p-değeri veya kritik bölge yaklaşımıyla karar verilir.
- Sonuç, probleme geri dönülerek yorumlanır (örneğin ortalama değişti mi?).


23. On dokuzuncu soruda elde edilen verinin normal dağılıma uyum uymadığını inceleyiniz ve yorumlayınız.

```r
set.seed(456)
norm_values <- rnorm(100, mean = 12, sd = 3)
shapiro_sonucu <- shapiro.test(norm_values)
shapiro_sonucu
qqnorm(norm_values)
qqline(norm_values, col = "red", lwd = 2)
# Çıktıda Shapiro-Wilk p-değeri ≈ 0.74 olduğundan (0.05'ten büyük), normal dağılım varsayımı reddedilmez.
```

24. On dokuzuncu sorudaki normal dağılım verisini 10 ortalama için test ediniz ve yorumlayınız.

```r
set.seed(456)
norm_values <- rnorm(100, mean = 12, sd = 3)
t_sonuc <- t.test(
    norm_values,
    mu = 10,
    alternative = "two.sided"
)
t_sonuc
# Beklenen çıktı: t ≈ 7.86 ve p ≈ 4.8e-12; p çok küçük olduğu için ortalama 10'dan anlamlı biçimde farklıdır.
```
